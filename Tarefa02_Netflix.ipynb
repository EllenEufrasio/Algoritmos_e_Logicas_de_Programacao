```python
# 12.Listar os 10 gêneros mais recorrentes.

# Assegura que o DataFrame df está carregado e com as colunas e tipos traduzidos
df = pd.read_csv("https://raw.githubusercontent.com/profzappa/profGit/refs/heads/master/netflix_titles.csv")

colunas_traduzidas = {
    'show_id': 'id_do_show',
    'type': 'tipo',
    'title': 'titulo',
    'director': 'diretor',
    'cast': 'elenco',
    'country': 'pais',
    'date_added': 'data_adicao',
    'release_year': 'ano_lancamento',
    'rating': 'classificacao',
    'duration': 'duracao',
    'listed_in': 'genero',
    'description': 'descricao'
}
df = df.rename(columns=colunas_traduzidas)

mapeamento_tipo = {
    'Movie': 'Filme',
    'TV Show': 'Série de TV'
}
df['tipo'] = df['tipo'].map(mapeamento_tipo)

# Mapeamento de gêneros individuais
mapeamento_generos = {
    'Dramas': 'Dramas',
    'International Movies': 'Filmes Internacionais',
    'Documentaries': 'Documentários',
    'Stand-Up Comedy': 'Comédia Stand-Up',
    'Comedies': 'Comédias',
    'Independent Movies': 'Filmes Independentes',
    'Kids\' TV': 'TV Infantil',
    'Children & Family Movies': 'Filmes Infantis e Família',
    'Romantic Movies': 'Filmes Românticos',
    'Action & Adventure': 'Ação e Aventura',
    'Thrillers': 'Suspenses',
    'Horror Movies': 'Filmes de Terror',
    'Music & Musicals': 'Música e Musicais',
    'Crime TV': 'Séries Policiais',
    'TV Dramas': 'Dramas de TV',
    'Teen TV Shows': 'Séries Adolescentes',
    'Sci-Fi & Fantasy': 'Ficção Científica e Fantasia',
    'British Movies': 'Filmes Britânicos',
    'Classic Movies': 'Filmes Clássicos',
    'Sports Movies': 'Filmes de Esporte',
    'Faith & Spirituality': 'Fé e Espiritualidade',
    'Anime Features': 'Filmes de Anime',
    'Anime Series': 'Séries de Anime',
    'LGBTQ Movies': 'Filmes LGBTQ+',
    'Korean Movies': 'Filmes Coreanos',
    'Spanish-Language Movies': 'Filmes em Espanhol',
    'Cult Movies': 'Filmes Cult',
    'TV Horror': 'TV de Terror',
    'TV Comedies': 'Comédias de TV',
    'TV Thrillers': 'Suspenses de TV',
    'TV Action & Adventure': 'TV de Ação e Aventura',
    'Docuseries': 'Documentários Seriados',
    'Reality TV': 'Reality Shows',
    'Science & Nature TV': 'TV de Ciência e Natureza',
    'Talk Shows & Stand-Up Comedy': 'Talk Shows e Comédia Stand-Up',
    'International TV Shows': 'Séries de TV Internacionais',
    'Movies': 'Filmes' # Adicionado para caso 'Movies' apareça sozinho
}

# Traduzir gêneros múltiplos
def traduzir_generos(generos_str):
    if isinstance(generos_str, str):
        generos = [g.strip() for g in generos_str.split(',')]
        generos_traduzidos = [mapeamento_generos.get(g, g) for g in generos]
        return ', '.join(generos_traduzidos)
    return generos_str # Retorna o valor original se não for string (ex: NaN)

# Aplica a tradução à coluna 'genero'
df['genero'] = df['genero'].apply(traduzir_generos)

print("Os 10 gêneros mais recorrentes (traduzidos):")
display(df['genero'].value_counts().head(10))
```

```python
# Para contar a ocorrência de cada gênero individualmente

# Primeiro, desmembrar as strings de gênero em uma lista de gêneros para cada entrada
all_genres = df['genero'].str.split(', ').explode()

# Em seguida, contar a frequência de cada gênero individual
genre_counts = all_genres.value_counts()

print("Contagem de gêneros individuais (TOP 10):")
display(genre_counts.head(10))
```
